# Lemmatization Demo

This notebook walks through how **lemmatization** works using spaCy.

We'll take a long piece of text, reduce every word to its lemma, and then check for discouraged words.

## Step 1: Download and load the spaCy model

Run this cell **before the session** to make sure the model is available.

```python
import spacy
from spacy.cli import download

# Download the small English model (skip if already installed)
download("en_core_web_sm")

# Load the model
nlp = spacy.load("en_core_web_sm")
print("Model loaded:", nlp.meta["name"], nlp.meta["version"])
```

## Step 2: Provide a sample text

Here's a longer passage that deliberately includes several forms of words.
Notice the mix of tenses, plurals, and conjugations.

```python
text = """
In this talk we delve into the rapidly shifting landscape of cloud-native
architectures. Attendees will be empowered with cutting-edge insights that
enhance their ability to navigate complex distributed systems. We equip
engineers with practical strategies for navigating microservice deployments,
enhancing observability, and delving deeper into resilience patterns.
The sessions are designed to empower teams and provide enhanced tooling
across every landscape they encounter.
"""

print(f"Text length: {len(text)} characters")
```

## Step 3: Tokenize and extract lemmas

spaCy processes the text into tokens. Each token carries a `.lemma_` attribute
that gives us the base (dictionary) form of the word.

For example:
- *navigating* -> *navigate*
- *enhanced* -> *enhance*
- *insights* -> *insight*

```python
doc = nlp(text)

# Show each token alongside its lemma
print(f"{'Token':<20} {'Lemma':<20}")
print("-" * 40)
for token in doc:
    if token.is_alpha:
        print(f"{token.text:<20} {token.lemma_:<20}")

# Collect unique lemmas (lowercase)
lemmas = {token.lemma_.lower() for token in doc if token.is_alpha}
print(f"\nUnique lemmas: {len(lemmas)}")
print(sorted(lemmas))
```

## Step 4: Filter out the offenders

We maintain a list of discouraged lemmas. These are words that tend to show up
in AI-generated or overly generic abstracts. If any of them appear in the
lemma set, we flag them.

```python
OFFENDERS = [
    "delve",
    "equip",
    "empower",
    "navigate",
    "landscape",
    "enhance",
    "insight",
]

found = sorted(word for word in OFFENDERS if word in lemmas)

if found:
    print("Offenders detected:")
    for word in found:
        print(f"  - {word}")
else:
    print("No offenders found.")
```
